```python
import speech_recognition as sr
from googletrans import Translator, LANGUAGES

def speech_to_text_audio(language='hi'):
    # Initialize recognizer and translator
    recognizer = sr.Recognizer()
    translator = Translator()

    # Capture the audio
    with sr.Microphone() as source:
        print("Please speak now...")
        audio = recognizer.listen(source)

        try:
            # Convert speech to text
            text = recognizer.recognize_google(audio)
            print(f"Original text: {text}")

            # Translate text to the selected language
            translated = translator.translate(text, dest=language)
            print(f"Translated text in {LANGUAGES[language]}: {translated.text}")

            # Optional: Convert translated text back into speech
            # (You can use a text-to-speech library like gTTS for this)

        except sr.UnknownValueError:
            print("Google Speech Recognition could not understand the audio")
        except sr.RequestError as e:
            print(f"Could not request results from Google Speech Recognition service; {e}")

# Example usage: Convert speech to French text
speech_to_text_audio('en')
```

```python
import os
import speech_recognition as sr
from googletrans import Translator, LANGUAGES
from gtts import gTTS

def speech_to_translated_speech(target_language='en'):
    # Initialize the recognizer and translator
    recognizer = sr.Recognizer()
    translator = Translator()

    with sr.Microphone() as source:
        print("Please speak now...")
        audio_data = recognizer.listen(source)

        try:
            # Speech recognition
            text = recognizer.recognize_google(audio_data)
            print(f"Recognized text: {text}")

            # Translation
            translated_text = translator.translate(text, dest=target_language).text
            print(f"Translated text ({LANGUAGES[target_language]}): {translated_text}")

            # Convert translated text to speech
            tts = gTTS(translated_text, lang=target_language)
            audio_file = "translated_speech.mp3"
            tts.save(audio_file)
            print(f"Translated speech saved as '{audio_file}'.")

            # Optionally, play the translated audio
            # For Linux/MacOS, you can use 'afplay'; for Windows, you can use 'start'
            os.system(f"afplay {audio_file}" if os.name == 'posix' else f"start {audio_file}")

        except sr.UnknownValueError:
            print("Could not understand the audio.")
        except sr.RequestError as e:
            print(f"Could not request results; {e}")

# Example usage: Convert spoken English to Spanish audio
speech_to_translated_speech('zh-cn')
```
